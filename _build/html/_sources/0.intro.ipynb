# Introduction
Traditional approaches to teaching statistics often view the discipline as a loosely-related collection of methods. The reality, however, is that the vast majority of statistical analyses are part of a singular generic framework. This framework hinges on the concept of a *statistical model*, where different analyses involve changing certain elements of the model, whilst leaving other elements the same. Viewing statistics this way allows for a deep appreciation of the *connection* between seemingly disparate analysis approaches. Furthermore, the steps involved in the analysis and the assumptions underlying the analysis are the *same*, irrespective of the specific model used. This allows for a *consolidation* of ideas into a single unifying framework that can be applied to a wide variety of datasets. Understanding this framework is one of the best things you can do to elevate your skills, allowing for a flexible and widely-applicable approach to statistical analysis.  

This lesson is about the core principles needed in order to fully appreciate this unifying framework. As such, it is vital that you study the ideas in this lesson as thoroughly as you can. This will necessarily be somewhat abstract at present, but the aim is to make this more concrete in the coming weeks as we see more examples. Because of this, we would encourage you to revisit this lesson periodically throughout the course as the more examples you are exposed to, the more this content will make sense. Learning is an iterative process, so we expect you to have a go at understanding everything in here, but recognise that much of it may only start to click once you see it in action. By the end of this unit, you should be able to revisit this lesson and understand it thoroughly.
